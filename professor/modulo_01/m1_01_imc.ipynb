**Pré-requisitos:** nenhum.

> **Versão do professor:** contém a solução de referência da atividade.

## Conceitos

- **Variável:** nome que guarda um valor. Use letras minúsculas, sem espaços, sem acentos e sem começar com número, como `preco_final`.
- **Tipos básicos:** `str` (texto), `int` (inteiro), `float` (decimal) e `bool` (`True` ou `False`).
- **`type()`:** mostra o tipo de um valor.
- **`input()`:** lê o que foi digitado e sempre retorna `str`.
- **`int()` e `float()`:** convertem um valor para número.
- **Operadores:** `+` (soma), `-` (subtração), `*` (multiplicação), `/` (divisão), `**` (potência), `//` (divisão inteira) e `%` (resto).
- **Ordem das operações:** `**` vem antes de `*`, `/`, `//` e `%`, que vêm antes de `+` e `-`. Parênteses mudam essa ordem.
- **`print()` e f-string:** `print()` mostra valores na saída. Em uma f-string (`f"..."`), o conteúdo entre `{}` é inserido no texto, e `{valor:.2f}` mostra o valor com 2 casas decimais.

## Exemplos

Rode as células em ordem, de cima para baixo.

```python
# Variáveis e tipos (dados fictícios)
ativo = "AAA"
quantidade = 100
preco = 38.50
comprado = True

print(type(ativo))       # <class 'str'>
print(type(quantidade))  # <class 'int'>
print(type(preco))       # <class 'float'>
print(type(comprado))    # <class 'bool'>
```

```python
# Operadores
print(7 / 2)        # 3.5
print(7 // 2)       # 3
print(7 % 2)        # 1
print(2 ** 3)       # 8
print(2 + 3 * 4)    # 14
print((2 + 3) * 4)  # 20
print(8 / 2 ** 2)   # 2.0
```

```python
# Texto x número
texto = "10"
numero = int(texto)

print(texto * 2)   # 1010
print(numero * 2)  # 20
```

```python
# input() e conversão
nome = input("Seu nome: ")
idade = int(input("Sua idade: "))

print(f"{nome} fará {idade + 1} anos no próximo aniversário.")
```

```python
# f-string com casas decimais
capital = 1000
taxa = 0.0125
juros = capital * taxa

print(f"Juros: R$ {juros:.2f}")  # Juros: R$ 12.50
```

## Erros comuns

| Erro | Exemplo | Causa |
|---|---|---|
| `TypeError` | `"10" + 1` | Soma de texto com número |
| `ValueError` | `float("1,75")` | Decimal escrito com vírgula |
| `NameError` | `print(precos)` | Nome digitado errado ou célula que define a variável ainda não executada |

## Atividade

Peça o peso (kg) e a altura (m), calcule `peso / altura ** 2` e imprima `IMC: X`, com 2 casas decimais. Digite a altura com ponto (ex.: `1.75`).

**Solução de referência**

```python
peso = float(input("Peso (kg): "))
altura = float(input("Altura (m): "))

imc = peso / altura ** 2

print(f"IMC: {imc:.2f}")
```
