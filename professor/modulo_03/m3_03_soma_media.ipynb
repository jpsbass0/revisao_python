# Módulo 3.3 — `while`, valor sentinela, soma e média

**Objetivos de aprendizagem**

- Repetir com `while` e `while True` + `break`.
- Usar valor sentinela para encerrar a leitura.
- Calcular soma e média com acumulador e contador, sem dividir por zero.

**Pré-requisitos:** Módulo 3.2.

> **Versão do professor:** contém a solução de referência da atividade.

## Conceitos

- **`while`:** repete enquanto a condição for verdadeira.
- **`while True`:** repete até encontrar um `break`. Serve quando não se sabe quantas repetições haverá.
- **Valor sentinela:** valor que encerra a leitura, sem entrar no cálculo.
- **Acumulador:** guarda a soma dos valores, com `total += valor`.
- **Média:** é a soma dividida pela quantidade, então você precisa de um acumulador e de um contador.
- **Divisão por zero:** gera `ZeroDivisionError`, então teste a quantidade antes de dividir.

## Exemplos

Rode as células em ordem, de cima para baixo.

```python
# while: precisa mudar algo a cada volta
contador = 1

while contador <= 3:
    print(contador)  # 1, 2, 3 (um por linha)
    contador += 1
```

```python
# while True com valor sentinela
quantidade = 0

while True:
    texto = input("Digite um texto (fim para sair): ")
    if texto == "fim":
        break
    quantidade += 1

print(quantidade)
```

```python
# Acumulador e contador
total = 0
quantidade = 0

for valor in [10, 20, 30]:
    total += valor
    quantidade += 1

print(total)             # 60
print(quantidade)        # 3
print(total / quantidade)  # 20.0
```

## Erros comuns

| Erro | Exemplo | Causa |
|---|---|---|
| `ZeroDivisionError` | `1 / 0` | Divisão sem testar a quantidade |
| `ValueError` | `float("abc")` | Texto que não representa número |
| Loop infinito | `input("Número: ") == 0` | `input()` retorna texto, e `"0" == 0` é `False` |
| Erro de lógica | Somar antes de testar o `0` | O `0` entra na contagem: `10, 20, 30, 0` dá média 15, não 20 |
| Erro de lógica | Testar `total == 0` no fim | `-5, 5, 0` tem soma 0, mas dois números informados |

> Se o loop não parar, clique em **Interrupt** no notebook.

## Atividade

Leia números até o usuário digitar `0`, que não entra no cálculo. Imprima `Soma: X` e, na linha seguinte, `Média: Y`, ambos com 2 casas decimais. Se o primeiro número for `0`, imprima `Nenhum número informado`. Digite decimais com ponto.

**Solução de referência**

```python
soma = 0
quantidade = 0

while True:
    numero = float(input("Número (0 para sair): "))
    if numero == 0:
        break
    soma += numero
    quantidade += 1

if quantidade == 0:
    print("Nenhum número informado")
else:
    print(f"Soma: {soma:.2f}")
    print(f"Média: {soma / quantidade:.2f}")
```
