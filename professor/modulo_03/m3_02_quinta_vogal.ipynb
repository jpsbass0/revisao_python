# Módulo 3.2 — `break` e variável de controle: a quinta vogal

**Objetivos de aprendizagem**

- Interromper um loop com `break`.
- Usar variável de controle para saber, depois do loop, se algo foi encontrado.
- Usar o valor de um contador para decidir quando parar.

**Pré-requisitos:** Módulo 3.1.

> **Versão do professor:** contém a solução de referência da atividade.

## Conceitos

- **`break`:** encerra o loop na hora. O código que vem depois do loop roda normalmente.
- **Variável de controle:** começa como `False` antes do loop e passa a `True` quando algo é encontrado. Depois do loop, ela diz se a busca deu certo.
- **`not`:** inverte um `bool`: `not True` é `False`, e `not False` é `True`.
- **Contador como critério de parada:** o contador do Módulo 3.1 pode ser testado dentro do loop, para parar quando chegar a um valor.

## Exemplos

Rode as células em ordem, de cima para baixo.

```python
# break: para no primeiro espaço
for letra in "Ana Maria":
    if letra == " ":
        break
    print(letra)  # A, n, a (uma por linha)

print("Fim")  # Depois do loop, o código continua
```

```python
# Variável de controle
palavra = "banana"
encontrou = False

for letra in palavra:
    if letra == "z":
        encontrou = True
        break

if not encontrou:
    print("Não encontrei")
```

```python
# not
print(not True)   # False
print(not False)  # True
```

## Erros comuns

| Erro | Exemplo | Causa |
|---|---|---|
| `NameError` | `encontrou` criada só dentro do `if` | Se nada for encontrado, a variável não existe depois do loop |
| Erro de lógica | `break` fora do `if` | O loop para na primeira volta |
| Erro de lógica | Imprimir o contador de vogais | A atividade pede a vogal, não a contagem |
| Erro de lógica | Testar `"Menos de 5 vogais"` dentro do loop | A mensagem aparece a cada volta |
| Erro de lógica | Testar a vogal sem `.lower()` | Com a entrada `AEIOU`, imprime `Menos de 5 vogais`, não `U` |

## Atividade

Leia uma frase e imprima a quinta vogal, como ela aparece na frase. Conte vogais maiúsculas e minúsculas. Se houver menos de cinco, imprima `Menos de 5 vogais`. Digite sem acentos.

**Solução de referência**

```python
frase = input("Digite uma frase: ")
vogais = "aeiou"
contagem_vogais = 0
encontrou = False

for letra in frase:
    if letra.lower() in vogais:
        contagem_vogais += 1
        if contagem_vogais == 5:
            encontrou = True
            print(letra)
            break

if not encontrou:
    print("Menos de 5 vogais")
```
