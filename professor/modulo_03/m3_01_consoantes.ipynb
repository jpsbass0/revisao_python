# Módulo 3.1 — Laço `for` e contagem de consoantes

**Objetivos de aprendizagem**

- Percorrer uma string com `for`.
- Usar contador, `in`/`not in` e `.lower()`.
- Distinguir o que roda a cada volta do que roda uma vez, ao final.

**Pré-requisitos:** Módulo 2.1.

> **Versão do professor:** contém a solução de referência da atividade.

## Conceitos

- **`for`:** repete o bloco para cada item de uma sequência. Em uma string, cada item é um caractere.
- **Contador:** começa em `0` antes do loop e aumenta com `+= 1`.
- **`len()`:** retorna o tamanho de uma string.
- **`in` e `not in`:** testam se um texto aparece ou não em outro.
- **`.lower()`:** retorna o texto em minúsculas, sem alterar a variável original.
- **Recuo:** o que está recuado repete a cada volta; o que está fora roda uma vez, ao final do loop.

## Exemplos

Rode as células em ordem, de cima para baixo.

```python
# for: uma letra por vez
for letra in "Ana":
    print(letra)  # A, n, a (uma por linha)
```

```python
# Contador com if dentro do for
contagem = 0

for letra in "Banana":
    if letra.lower() in "aeiou":
        contagem += 1

print(contagem)  # 3
```

```python
# Tamanho, in e lower()
print(len("banana"))        # 6
print("a" in "aeiou")       # True
print("A" in "aeiou")       # False
print("A".lower())          # a
print("b" not in "aeiou")   # True
```

## Erros comuns

| Erro | Exemplo | Causa |
|---|---|---|
| `TypeError` | `1 in "aeiou"` | Busca de número em texto |
| `TypeError` | `for x in 123:` | Número não pode ser percorrido |
| `NameError` | `contagem += 1` sem criar `contagem` | Contador não iniciado antes do loop |
| Erro de lógica | `print()` recuado dentro do `for` | Imprime a contagem a cada volta, e não só o total ao final |
| Erro de lógica | `not in "aeiou"` com a entrada `Ana Maria` | Conta o espaço como consoante: dá 4, não 3 |

## Atividade

Leia uma palavra e imprima quantas consoantes ela tem, contando maiúsculas e minúsculas. Digite sem acento.

**Solução de referência**

```python
palavra = input("Digite uma palavra: ")
consoantes = "bcdfghjklmnpqrstvwxyz"
contagem_consoantes = 0

for letra in palavra:
    if letra.lower() in consoantes:
        contagem_consoantes += 1

print(contagem_consoantes)
```
