# Módulo 3.1 — Laço `for`: as consoantes de uma frase

**Objetivos de aprendizagem**

- Percorrer uma string com `for`.
- Usar contador, acumulador de texto, `in`/`not in` e `.lower()`.
- Distinguir o que roda a cada volta do que roda uma vez, ao final.

**Pré-requisitos:** Módulo 2.1.

> **Versão do professor:** contém a solução de referência da atividade.

## Conceitos

- **`for`:** repete o bloco para cada item de uma sequência. Em uma string, cada item é um caractere.
- **Contador:** começa em `0` antes do loop e aumenta com `+= 1`.
- **Acumulador de texto:** começa como texto vazio (`""`) antes do loop e recebe caracteres com `+=`.
- **`in` e `not in`:** testam se um texto aparece ou não em outro.
- **`.lower()`:** retorna o texto em minúsculas, sem alterar a variável original.
- **Recuo:** o que está recuado repete a cada volta; o que está fora roda uma vez, ao final do loop.

## Exemplos

Rode as células em ordem, de cima para baixo.

```python
# for: uma letra por vez
for letra in "Ana":
    print(letra)  # A, n, a (uma por linha)
```

```python
# Contador com if dentro do for
contagem = 0

for letra in "Banana":
    if letra.lower() in "aeiou":
        contagem += 1

print(contagem)  # 3
```

```python
# Acumulador de texto: guarda só os dígitos
valor = "R$ 1.250,00"
digitos = ""

for caractere in valor:
    if caractere in "0123456789":
        digitos += caractere

print(digitos)  # 125000
```

```python
# in, not in e lower()
print("a" in "aeiou")       # True
print("A" in "aeiou")       # False
print("A".lower())          # a
print("b" not in "aeiou")   # True
```

## Erros comuns

| Erro | Exemplo | Causa |
|---|---|---|
| `TypeError` | `1 in "aeiou"` | Busca de número em texto |
| `TypeError` | `for x in 123:` | Número não pode ser percorrido |
| `NameError` | `resultado += letra` sem criar `resultado` | Acumulador não iniciado antes do loop |
| Erro de lógica | `print()` recuado dentro do `for` | Imprime o resultado parcial a cada volta, e não só o final |
| Erro de lógica | `not in "aeiou"` com a entrada `Ana Maria` | Inclui o espaço: imprime `n Mr`, não `nMr` |

## Atividade

Leia uma frase e imprima apenas as consoantes, na ordem em que aparecem e em uma única linha. Mantenha maiúsculas e minúsculas como estão na frase; espaços e pontuação não são consoantes. Digite sem acentos e sem `ç`.

**Solução de referência**

```python
frase = input("Digite uma frase: ")
consoantes = "bcdfghjklmnpqrstvwxyz"
resultado = ""

for letra in frase:
    if letra.lower() in consoantes:
        resultado += letra

print(resultado)
```
