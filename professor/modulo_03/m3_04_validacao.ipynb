# Módulo 3.4 — Validação de entrada com `while`

**Objetivos de aprendizagem**

- Repetir uma pergunta enquanto a resposta for inválida.
- Combinar condições com `and` e `or`.
- Aplicar a leitura dupla (antes e dentro do loop).

**Pré-requisitos:** Módulo 3.3.

> **Versão do professor:** contém a solução de referência da atividade.

## Conceitos

- **Validação com `while`:** repete a pergunta enquanto a resposta for inválida.
- **Condição do `while`:** descreve o valor inválido, não o válido.
- **`and`:** é verdadeiro só se as duas condições forem verdadeiras.
- **`or`:** é verdadeiro se pelo menos uma condição for verdadeira.
- **Leitura dupla:** o primeiro `input()` vem antes do loop, e o segundo fica dentro dele. Sem o segundo, o valor nunca muda e o loop não termina.

## Exemplos

Rode as células em ordem, de cima para baixo.

```python
# and e or
idade = 20

print(idade >= 18 and idade < 65)  # True
print(idade < 18 or idade >= 65)   # False
```

```python
# Validação com while
resposta = input("Continuar? (s/n): ")

while resposta != "s" and resposta != "n":
    print("Resposta inválida.")
    resposta = input("Continuar? (s/n): ")

print(f"Você respondeu: {resposta}")
```

## Erros comuns

| Erro | Exemplo | Causa |
|---|---|---|
| `SyntaxError` | `if x < 0 or > 10:` | Falta repetir a variável depois do `or` |
| `ValueError` | `float("7,5")` | Decimal escrito com vírgula |
| Loop infinito | `input()` só antes do `while` | O valor nunca muda |
| Loop infinito | `resposta != "s" or resposta != "n"` | Sempre verdadeiro, porque a resposta nunca é igual às duas ao mesmo tempo |

## Atividade

Peça uma nota até o usuário digitar um valor de 0 a 10. Depois, imprima `Nota válida: X`. Digite decimais com ponto.

**Solução de referência**

```python
nota = float(input("Nota (0 a 10): "))

while nota < 0 or nota > 10:
    nota = float(input("Nota (0 a 10): "))

print(f"Nota válida: {nota}")
```
