**Pré-requisitos:** Módulo 3.4.

## Conceitos

- **Módulo do Python:** arquivo com funções prontas; não confundir com os módulos desta trilha. `import random` dá acesso às funções do módulo `random`, que são chamadas com o prefixo `random.`.
- **`random.randint(a, b)`:** sorteia um inteiro entre `a` e `b`, incluindo os dois extremos. O resultado muda a cada execução.
- **Onde fica o `import`:** no início do código, uma vez só. Em um notebook, depois que a célula com o `import` roda, o módulo vale para as células seguintes.
- **Loop com limite:** `while usadas < tentativas:` repete no máximo `tentativas` vezes, desde que `usadas` aumente a cada volta.
- **Duas saídas do loop:** o `break` encerra o loop no acerto; sem acerto, o loop termina quando a condição fica falsa. A variável de controle (Módulo 3.2) diz, depois do loop, qual das duas aconteceu.

## Exemplos

Rode as células em ordem, de cima para baixo.

```python
# Sorteio: o resultado muda a cada execução
import random

dado = random.randint(1, 6)
print(dado)
```

```python
# Os extremos entram no sorteio: com início igual ao fim, só há um resultado possível
print(random.randint(5, 5))  # 5
```

```python
# Loop com no máximo 3 voltas
usadas = 0

while usadas < 3:
    usadas += 1
    print(f"Volta {usadas}")

print("Fim")
```

```python
# Duas saídas: acerto (break) ou fim das tentativas
senha = "python"
usadas = 0
acertou = False

while usadas < 3:
    palpite = input("Senha: ")
    usadas += 1
    if palpite == senha:
        acertou = True
        break

if acertou:
    print("Acesso liberado")
else:
    print("Tentativas esgotadas")
```

## Erros comuns

| Erro | Exemplo | Causa |
|---|---|---|
| `NameError` | `randint(1, 6)` | A função pertence ao módulo; escreva `random.randint` |
| `NameError` | `random.randint(1, 6)` sem `import random` | O módulo não foi importado, ou a célula do `import` não foi rodada |
| `ValueError` | `random.randint(10, 1)` | O início da faixa é maior que o fim |
| Erro de lógica | `if chute == sorteado:` com `chute = input("Seu chute: ")` | `input()` retorna `str`, e `"5" == 5` é `False`; converta com `int()` |
| Erro de lógica | Sortear dentro do loop | O número muda a cada chute |
| Loop infinito | Esquecer `usadas += 1` | A condição nunca fica falsa |

> Se o loop não parar, clique em **Interrupt** no notebook.

## Atividade

Jogo da adivinhação: peça o início e o fim da faixa de valores e a quantidade de tentativas possíveis. O computador sorteia um número inteiro na faixa indicada, incluindo os extremos, e o usuário tenta adivinhar. A cada chute errado, imprima `Errou.`. O jogo acaba se o usuário adivinhar, com a mensagem `Acertou!`, ou se terminarem as tentativas possíveis, com a mensagem `Fim das tentativas. O número era X`.

Considere que todas as entradas são números inteiros e que o início da faixa não é maior que o fim.

**Solução de referência**

```python
import random

inicio = int(input("Início da faixa: "))
fim = int(input("Fim da faixa: "))
tentativas = int(input("Quantidade de tentativas: "))

sorteado = random.randint(inicio, fim)
usadas = 0
acertou = False

while usadas < tentativas:
    chute = int(input("Seu chute: "))
    usadas += 1
    if chute == sorteado:
        acertou = True
        break
    print("Errou.")

if acertou:
    print("Acertou!")
else:
    print(f"Fim das tentativas. O número era {sorteado}")
```
