- Comparar elementos por um critério: a quantidade de caracteres.
- Decidir o que fazer em caso de empate.

**Pré-requisitos:** Módulo 5.1.

> **Versão do professor:** contém a solução de referência da atividade.

## Conceitos

- **`len()` em string:** retorna a quantidade de caracteres, contando espaços e pontuação. `len("sol")` é 3.
- **Comparar por um critério:** guarde o elemento (`maior`), mas compare o tamanho (`len(palavra) > len(maior)`).
- **Empate:** com `>`, fica o primeiro elemento de maior tamanho; com `>=`, o último o substitui.
- **`>` entre strings:** compara a ordem alfabética, não o tamanho. `"zebra" > "abacaxi"` é `True`.

## Exemplos

Rode as células em ordem, de cima para baixo.

```python
# len() conta caracteres, inclusive espaços
print(len("sol"))             # 3
print(len("Rio de Janeiro"))  # 14
print(len(""))                # 0
```

```python
# > entre strings compara a ordem alfabética, não o tamanho
print("zebra" > "abacaxi")            # True
print(len("zebra") > len("abacaxi"))  # False
```

```python
# Menor string: guarda o elemento, compara o tamanho
cidades = ["Recife", "Rio", "Salvador"]
menor = cidades[0]

for cidade in cidades:
    if len(cidade) < len(menor):
        menor = cidade

print(menor)  # Rio
```

## Erros comuns

| Erro | Exemplo | Causa |
|---|---|---|
| Erro de lógica | `if palavra > maior:` | Compara a ordem alfabética, não o tamanho |
| Erro de lógica | `maior = len(palavra)` | Guarda o tamanho e perde a palavra |
| Erro de lógica | `>=` no lugar de `>` | Em empate, fica com o último elemento, não com o primeiro |
| `TypeError` | `len(5)` | Número não tem tamanho; a lista deve ter só strings |

## Atividade

Imprima o elemento com maior quantidade de caracteres de uma lista que contenha apenas elementos do tipo string, no formato `Maior string: X`. Em caso de empate, imprima o primeiro deles. Percorra a lista com `for`.

Para testar, troque a lista na primeira linha da célula.

**Solução de referência**

```python
palavras = ["sol", "abacaxi", "mar"]
maior = palavras[0]

for palavra in palavras:
    if len(palavra) > len(maior):
        maior = palavra

print(f"Maior string: {maior}")
```
