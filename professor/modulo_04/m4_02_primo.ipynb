**Objetivos de aprendizagem**

- Reaproveitar a contagem de divisores do Módulo 4.1.
- Tomar uma decisão com base no resultado de um loop.
- Testar casos-limite, como o número 1.

**Pré-requisitos:** Módulo 4.1.

> **Versão do professor:** contém a solução de referência da atividade.

## Conceitos

- **Número primo:** tem exatamente dois divisores positivos, 1 e ele mesmo. Exemplos: 2, 3, 5, 7 e 11.
- **O número 1:** tem um único divisor, então não é primo.
- **Decidir depois do loop:** o `if` que usa a contagem fica fora do `for`, porque a contagem só está completa quando o loop termina.
- **Reaproveitar código:** a contagem de divisores do Módulo 4.1 resolve a primeira parte do problema; falta comparar o resultado com 2.
- **Caso-limite:** valor na fronteira de uma regra, como 1 e 2 aqui. Vale testá-lo sempre.

## Exemplos

Rode as células em ordem, de cima para baixo.

```python
# Contar e decidir depois do loop
numero = 10
multiplos_de_3 = 0

for i in range(1, numero + 1):
    if i % 3 == 0:
        multiplos_de_3 += 1

if multiplos_de_3 > 2:
    print("Mais de dois múltiplos de 3")
else:
    print("No máximo dois múltiplos de 3")
```

```python
# Listar os divisores, um por linha
numero = 6

for i in range(1, numero + 1):
    if numero % i == 0:
        print(i)
```

```python
# Uma comparação com == já resulta em bool
divisores = 2

print(divisores == 2)  # True
```

## Erros comuns

| Erro | Exemplo | Causa |
|---|---|---|
| Erro de lógica | `if divisores == 2:` dentro do `for` | Decide antes de contar todos os divisores |
| Erro de lógica | `if divisores <= 2:` | Classifica o 1 como primo |
| Erro de lógica | `divisores = 0` dentro do `for` | A contagem volta a zero a cada volta |
| Erro de lógica | `range(2, numero)` com o teste `divisores == 2` | Essa faixa exclui 1 e o próprio número, então a contagem de um primo dá 0 |

## Atividade

A partir da contagem de seus divisores, informe se um número inteiro positivo é primo. Imprima no formato `7 é primo` ou `12 não é primo`.

**Solução de referência**

```python
numero = int(input("Número inteiro positivo: "))
divisores = 0

for i in range(1, numero + 1):
    if numero % i == 0:
        divisores += 1

if divisores == 2:
    print(f"{numero} é primo")
else:
    print(f"{numero} não é primo")
```
