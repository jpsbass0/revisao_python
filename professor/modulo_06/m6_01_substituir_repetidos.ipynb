# Módulo 6.1 — Posições em strings: substituir repetições por `@`

**Objetivos de aprendizagem**

- Acessar caracteres pela posição, com `texto[i]`.
- Percorrer as posições com `range(len(texto))`.
- Comparar cada caractere com o anterior.

**Pré-requisitos:** Módulo 5.3.

> **Versão do professor:** contém a solução de referência da atividade.

## Conceitos

- **Posição em string:** funciona como em lista. `texto[0]` é o primeiro caractere; `texto[-1]`, o último.
- **`range(len(texto))`:** gera as posições válidas, de 0 a `len(texto) - 1`.
- **Caractere anterior:** na posição `i`, o anterior é `texto[i - 1]`. Na posição 0 não há anterior: `texto[0 - 1]` é `texto[-1]`, o último caractere. Por isso, teste `i > 0` antes de comparar.
- **Curto-circuito do `and`:** em `i > 0 and texto[i] == texto[i - 1]`, a comparação só roda quando `i > 0` (Módulo 3.4).
- **Strings não podem ser alteradas:** `texto[0] = "x"` gera erro. Para obter um texto modificado, monte outro com um acumulador.

## Exemplos

Rode as células em ordem, de cima para baixo.

```python
# Posições
palavra = "banana"

print(palavra[0])    # b
print(palavra[-1])   # a
print(len(palavra))  # 6
```

```python
# range(len()) gera as posições
for i in range(len(palavra)):
    print(i, palavra[i])
```

```python
# Comparar com o anterior: em que posições o caractere repete o de antes?
texto = "aabccc"

for i in range(len(texto)):
    if i > 0 and texto[i] == texto[i - 1]:
        print(f"Repetição na posição {i}")
```

```python
# Montar um texto novo: troca espaços por _
frase = "Ola mundo"
resultado = ""

for i in range(len(frase)):
    if frase[i] == " ":
        resultado += "_"
    else:
        resultado += frase[i]

print(resultado)  # Ola_mundo
```

## Erros comuns

| Erro | Exemplo | Causa |
|---|---|---|
| `IndexError` | `texto[len(texto)]` | A última posição é `len(texto) - 1` |
| `TypeError` | `texto[0] = "x"` | Strings não podem ser alteradas; monte um texto novo |
| Erro de lógica | `texto[i] == texto[i - 1]` sem testar `i > 0` | Na posição 0, compara o primeiro caractere com o último |
| `IndexError` | `texto[i + 1]` na última posição | Não há próximo caractere |
| `TypeError` | `for i in range(texto):` | `range()` precisa de um número; use `range(len(texto))` |

## Atividade

Leia uma string e imprima-a substituindo as sequências de caracteres repetidos por `@`: o primeiro caractere de cada sequência fica, e cada repetição seguinte vira `@`. Por exemplo, `abcdddxxmmmmaymm` dará como resultado `abcd@@x@m@@@aym@`. Maiúsculas e minúsculas são caracteres diferentes.

**Solução de referência**

```python
texto = input("Digite uma string: ")
resultado = ""

for i in range(len(texto)):
    if i > 0 and texto[i] == texto[i - 1]:
        resultado += "@"
    else:
        resultado += texto[i]

print(resultado)
```
