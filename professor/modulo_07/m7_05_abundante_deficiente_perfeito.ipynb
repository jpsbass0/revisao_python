## Conceitos

- **Divisores próprios:** todos os divisores positivos de `n`, menos o próprio `n`. Os de 12 são 1, 2, 3, 4 e 6.
- **Soma dos divisores próprios:** é `soma_divisores(n) - n`, usando a função do Módulo 7.4.
- **Classificação:** compare essa soma com `n`.
  - Igual: o número é **perfeito** (`"P"`). Exemplos: 6, porque 1 + 2 + 3 = 6, e 28.
  - Maior: o número é **abundante** (`"A"`). Exemplo: 12, porque 1 + 2 + 3 + 4 + 6 = 16.
  - Menor: o número é **deficiente** (`"D"`). Exemplos: todo primo, como 7, cuja soma é 1, e o 1, cuja soma é 0.
- **Função que usa outra:** dentro de `classificar_numero`, chame `soma_divisores`. A função chamada precisa estar definida antes de a chamada rodar.
- **Fonte das definições:** [Número perfeito, na Wikipédia](https://pt.wikipedia.org/wiki/N%C3%BAmero_perfeito), indicada na lista de exercícios.

## Exemplos

Rode as células em ordem, de cima para baixo.

```python
# Uma função usando outra
def quadrado(x):
    return x * x

def soma_dos_quadrados(a, b):
    return quadrado(a) + quadrado(b)

print(soma_dos_quadrados(3, 4))  # 25
```

```python
# Comparar um resultado com uma referência e retornar um código
def saldo(receitas, despesas):
    return receitas - despesas

def situacao_caixa(receitas, despesas):
    resultado = saldo(receitas, despesas)
    if resultado > 0:
        return "Superávit"
    elif resultado < 0:
        return "Déficit"
    else:
        return "Equilíbrio"

print(situacao_caixa(100, 80))  # Superávit
print(situacao_caixa(50, 50))   # Equilíbrio
```

## Erros comuns

| Erro | Exemplo | Causa |

## Atividade

Crie uma função `classificar_numero(n)` que receba um número inteiro positivo e retorne um dos três possíveis valores: `"A"`, `"D"` ou `"P"`, dependendo se o número informado é abundante, deficiente ou perfeito. Use o resultado da função `soma_divisores` do Módulo 7.4. Depois, peça um número e imprima no formato `12: A`.

**Solução de referência**

```python
def soma_divisores(n):
    soma = 0
    for i in range(1, n + 1):
        if n % i == 0:
            soma += i
    return soma

def classificar_numero(n):
    soma_proprios = soma_divisores(n) - n
    if soma_proprios > n:
        return "A"
    elif soma_proprios < n:
        return "D"
    else:
        return "P"

numero = int(input("Número inteiro positivo: "))
print(f"{numero}: {classificar_numero(numero)}")
```
