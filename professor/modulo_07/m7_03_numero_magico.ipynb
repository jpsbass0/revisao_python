# Módulo 7.3 — Encapsular código: o Jogo do Número Mágico

**Objetivos de aprendizagem**

- Encapsular um programa inteiro em uma função sem parâmetros.
- Chamar a mesma função várias vezes dentro de um loop.
- Repetir o jogo até o usuário decidir parar.

**Pré-requisitos:** Módulos 3.4, 3.5 e 7.2.

> **Versão do professor:** contém a solução de referência da atividade.

## Conceitos

- **Encapsular:** colocar um trecho de código dentro de uma função. O código é o mesmo, agora recuado sob o `def`.
- **Função sem parâmetros:** `def jogar():` não recebe valores; ela mesma lê o que precisa com `input()`. A chamada mantém os parênteses: `jogar()`.
- **Função sem `return`:** quando o objetivo é mostrar mensagens, a função pode usar só `print()`. Ela devolve `None`.
- **Repetir até o usuário decidir parar:** um `while True` chama a função e, depois, pergunta se o usuário quer jogar de novo; a resposta `n` leva ao `break`.
- **Loop dentro de loop:** a validação da resposta (Módulo 3.4) é um `while True` dentro do `while True` das partidas. O `break` encerra só o loop que o contém diretamente.
- **`import` no início:** importe `random` uma vez, antes do `def`.

## Exemplos

Rode as células em ordem, de cima para baixo.

```python
# Encapsular: o mesmo código, agora com nome
def saudacao():
    nome = input("Seu nome: ")
    print(f"Olá, {nome}!")

saudacao()
```

```python
# A mesma função chamada a cada volta de um loop
def linha():
    print("----------")

for i in range(3):
    linha()
```

```python
# Repetir até o usuário decidir parar
while True:
    saudacao()
    resposta = input("De novo? (s/n): ")
    if resposta == "n":
        break

print("Fim")
```

```python
# O break encerra só o loop que o contém diretamente
for rodada in range(2):
    while True:
        print(f"Rodada {rodada}")
        break
    print("Continua no for")
```

## Erros comuns

| Erro | Exemplo | Causa |
|---|---|---|
| Erro de lógica | `jogar` sem parênteses | O nome sozinho não chama a função; nada acontece |
| `NameError` | Chamar `jogar()` antes de rodar o `def` | A função ainda não foi definida |
| Erro de lógica | Parte do jogo sem recuo sob o `def` | Esse trecho fica fora da função e não roda a cada partida |
| Loop infinito | Perguntar `Jogar de novo?` antes do `while True` | A resposta é lida uma vez só e nunca muda |
| Erro de lógica | Esperar que o `break` da validação encerre as partidas | Ele encerra só o `while True` interno |

> Se o loop não parar, clique em **Interrupt** no notebook.

## Atividade

Encapsule em uma função `jogar()` o código do Jogo do Número Mágico, o jogo da adivinhação do Módulo 3.5, com as mesmas entradas e mensagens. Com esse código encapsulado, crie o código necessário para que o jogo seja jogado repetidamente até que o usuário decida parar: depois de cada partida, pergunte `Jogar de novo? (s/n)`, aceitando só `s` ou `n` e perguntando de novo se a resposta for outra. Ao parar, imprima `Fim de jogo.`.

Considere que as entradas do jogo são números inteiros e que o início da faixa não é maior que o fim.

**Solução de referência**

```python
import random

def jogar():
    inicio = int(input("Início da faixa: "))
    fim = int(input("Fim da faixa: "))
    tentativas = int(input("Quantidade de tentativas: "))

    sorteado = random.randint(inicio, fim)
    usadas = 0
    acertou = False

    while usadas < tentativas:
        chute = int(input("Seu chute: "))
        usadas += 1
        if chute == sorteado:
            acertou = True
            break
        print("Errou.")

    if acertou:
        print("Acertou!")
    else:
        print(f"Fim das tentativas. O número era {sorteado}")

while True:
    jogar()
    while True:
        resposta = input("Jogar de novo? (s/n): ")
        if resposta == "s" or resposta == "n":
            break
    if resposta == "n":
        break

print("Fim de jogo.")
```
