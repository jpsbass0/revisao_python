# Módulo 7.4 — Função com loop: a soma dos divisores

**Objetivos de aprendizagem**

- Escrever uma função com loop e acumulador.
- Retornar o resultado só depois do loop.
- Separar o cálculo, feito na função, da leitura e da exibição.

**Pré-requisitos:** Módulos 4.1 e 7.3.

> **Versão do professor:** contém a solução de referência da atividade.

## Conceitos

- **Acumulador dentro da função:** a soma começa em `0` dentro da função, para recomeçar a cada chamada.
- **`return` depois do loop:** o `return` fica fora do `for`, no recuo do corpo da função. Dentro do loop, ele encerraria a função na primeira volta.
- **Divisores de `n`:** os mesmos do Módulo 4.1: os `i` entre 1 e `n` com `n % i == 0`. A soma inclui 1 e o próprio `n`.
- **Separar cálculo e exibição:** a função só calcula e retorna; o `input()` e o `print()` ficam fora dela. Assim, ela pode ser reaproveitada, como no Módulo 7.5.

## Exemplos

Rode as células em ordem, de cima para baixo.

```python
# Acumulador dentro da função
def soma_ate(n):
    soma = 0
    for i in range(1, n + 1):
        soma += i
    return soma

print(soma_ate(4))    # 10
print(soma_ate(100))  # 5050
```

```python
# return dentro do loop: este código está errado de propósito
def soma_errada(n):
    soma = 0
    for i in range(1, n + 1):
        soma += i
        return soma

print(soma_errada(4))  # 1
```

```python
# O resultado da função pode entrar em outra conta
def quantidade_impares(n):
    quantidade = 0
    for i in range(1, n + 1):
        if i % 2 == 1:
            quantidade += 1
    return quantidade

print(quantidade_impares(10) * 2)  # 10
```

## Erros comuns

| Erro | Exemplo | Causa |
|---|---|---|
| Erro de lógica | `return` dentro do `for` | A função termina na primeira volta |
| `UnboundLocalError` | `soma = 0` fora da função e `soma += i` dentro | Dentro da função, `soma` é local e ainda não tem valor quando é somada |
| Erro de lógica | `range(1, n)` | Deixa o próprio `n` de fora da soma |
| Erro de lógica | `print(soma)` no lugar de `return soma` | A função devolve `None`, e o Módulo 7.5 não consegue usar o resultado |

## Atividade

Crie uma função `soma_divisores(n)` que receba um número inteiro positivo e calcule a soma dos seus divisores, incluindo 1 e o próprio número, retornando o resultado. Depois, peça um número, chame a função e imprima no formato `Soma dos divisores de 12: 28`.

**Solução de referência**

```python
def soma_divisores(n):
    soma = 0
    for i in range(1, n + 1):
        if n % i == 0:
            soma += i
    return soma

numero = int(input("Número inteiro positivo: "))
print(f"Soma dos divisores de {numero}: {soma_divisores(numero)}")
```
