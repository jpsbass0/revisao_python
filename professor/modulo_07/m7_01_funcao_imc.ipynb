# Módulo 7.1 — Funções: `def`, parâmetros e `return`

**Objetivos de aprendizagem**

- Definir funções com `def` e chamá-las.
- Receber valores por parâmetros e devolver o resultado com `return`.
- Distinguir `return` de `print()`.

**Pré-requisitos:** Módulo 6.4.

> **Versão do professor:** contém a solução de referência da atividade.

## Conceitos

- **Função:** bloco de código com nome, que pode ser chamado várias vezes. Você já usou funções prontas, como `len()` e `input()`.
- **`def`:** define a função: `def nome(parametro1, parametro2):`, seguido do bloco recuado.
- **Parâmetros:** variáveis que recebem os valores passados na chamada. Em `area_retangulo(3, 4)`, `3` vai para o primeiro parâmetro, e `4`, para o segundo.
- **`return`:** encerra a função e devolve um valor a quem a chamou. Sem `return`, a função devolve `None`.
- **`return` x `print()`:** `print()` só mostra o valor na tela; `return` entrega o valor, que pode ser guardado em uma variável ou usado em outra conta.
- **Definir não é executar:** o bloco da função só roda quando ela é chamada.
- **Variáveis locais:** variáveis criadas dentro da função só existem dentro dela.

## Exemplos

Rode as células em ordem, de cima para baixo.

```python
# Definir e chamar
def area_retangulo(base, altura):
    return base * altura

print(area_retangulo(3, 4))  # 12
resultado = area_retangulo(2.5, 2)
print(resultado)             # 5.0
```

```python
# return devolve o valor; print() só mostra
def dobro_com_print(x):
    print(x * 2)

def dobro_com_return(x):
    return x * 2

a = dobro_com_print(5)   # mostra 10 na tela
b = dobro_com_return(5)  # não mostra nada
print(a)  # None
print(b)  # 10
```

```python
# Juros simples: função com três parâmetros
def juros_simples(capital, taxa, meses):
    return capital * taxa * meses

juros = juros_simples(1000, 0.01, 12)
print(f"Juros: R$ {juros:.2f}")  # Juros: R$ 120.00
```

```python
# Variável local: só existe dentro da função
def mensagem():
    texto = "Olá"
    return texto

print(mensagem())  # Olá
```

## Erros comuns

| Erro | Exemplo | Causa |
|---|---|---|
| `NameError` | Chamar a função antes de rodar a célula do `def` | A função ainda não foi definida |
| `TypeError` | `area_retangulo(3)` | Falta um argumento: a função tem dois parâmetros |
| Erro de lógica | `print()` no lugar de `return` | A função mostra o valor, mas devolve `None` |
| `NameError` | Usar fora da função uma variável criada dentro dela | Variáveis locais só existem dentro da função |
| Erro de lógica | Código depois do `return` | Nunca roda, porque o `return` encerra a função |
| `IndentationError` | Bloco da função sem recuo | O recuo define o que pertence à função |

## Atividade

Crie uma função `calcular_imc(peso, altura)` que receba o peso (kg) e a altura (m) de uma pessoa e retorne o seu índice IMC. Depois, peça o peso e a altura, chame a função e imprima `IMC: X`, com 2 casas decimais. Digite a altura com ponto (ex.: `1.75`).

**Solução de referência**

```python
def calcular_imc(peso, altura):
    return peso / altura ** 2

peso = float(input("Peso (kg): "))
altura = float(input("Altura (m): "))
imc = calcular_imc(peso, altura)
print(f"IMC: {imc:.2f}")
```
