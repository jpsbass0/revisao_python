# Módulo 5.3 — Construir listas: separar strings e números

**Objetivos de aprendizagem**

- Criar uma lista vazia e acrescentar elementos com `.append()`.
- Identificar o tipo de um valor com `type()`.
- Distribuir os elementos de uma lista em duas listas.

**Pré-requisitos:** Módulo 5.2.

## Conceitos

- **Lista vazia:** `[]`. É o ponto de partida, como o `0` de um contador ou o `""` de um acumulador de texto.
- **`.append(x)`:** acrescenta `x` ao final da própria lista. Não escreva `lista = lista.append(x)`: o `.append()` retorna `None`, e a lista se perde.
- **`type()`:** visto no Módulo 1.1. `type("a") == str` é `True`, assim como `type(3) == int` e `type(4.5) == float`.
- **Separar em dois grupos:** com `if`/`else`, cada elemento vai para exatamente uma das listas.
- **`print()` de lista:** mostra os elementos entre colchetes, com as strings entre aspas: `['ana', 'bia']`.

## Exemplos

Rode as células em ordem, de cima para baixo.

```python
# Lista vazia e append()
compras = []
compras.append("arroz")
compras.append("feijão")

print(compras)       # ['arroz', 'feijão']
print(len(compras))  # 2
```

```python
# type() identifica o tipo
print(type("ana") == str)  # True
print(type(3) == str)      # False
print(type(4.5) == float)  # True
```

```python
# Separar pares e ímpares
numeros = [4, 7, 10, 3]
pares = []
impares = []

for n in numeros:
    if n % 2 == 0:
        pares.append(n)
    else:
        impares.append(n)

print(pares)    # [4, 10]
print(impares)  # [7, 3]
```

## Erros comuns

| Erro | Exemplo | Causa |
|---|---|---|
| `AttributeError` | `lista = lista.append(1)` seguido de `lista.append(2)` | O primeiro `.append()` retorna `None`, que passa a ocupar a variável |
| Erro de lógica | `strings = []` dentro do `for` | A lista volta a ficar vazia a cada volta |
| Erro de lógica | `if type(x) == "str":` | Compara com o texto `"str"`, e não com o tipo `str`; o resultado é sempre `False` |
| `TypeError` | `strings.append()` | Falta o elemento a acrescentar |
