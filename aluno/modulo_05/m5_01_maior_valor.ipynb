# Módulo 5.1 — Listas: o elemento de maior valor

**Objetivos de aprendizagem**

- Criar listas e acessar elementos pela posição.
- Percorrer uma lista com `for`.
- Guardar o maior valor encontrado até o momento.

**Pré-requisitos:** Módulo 4.2.

## Conceitos

- **Lista:** sequência de valores entre colchetes, separados por vírgula: `[3, 17, -2]`. Pode misturar tipos.
- **Posição (índice):** começa em 0. `lista[0]` é o primeiro elemento; `lista[-1]`, o último.
- **`len()`:** retorna a quantidade de elementos.
- **`for` em lista:** `for x in lista:` visita cada elemento, em ordem, como em uma string.
- **Maior até o momento:** comece com o primeiro elemento (`maior = lista[0]`) e troque sempre que achar um valor maior. Começar com `0` falha quando todos os valores são negativos.
- **`max()`:** o Python tem essa função pronta. Na atividade, não a use: o objetivo é praticar o percurso da lista.

## Exemplos

Rode as células em ordem, de cima para baixo.

```python
# Lista, posição e tamanho
precos = [12.5, 8.0, 30.0, 4.75]

print(precos[0])    # 12.5
print(precos[-1])   # 4.75
print(len(precos))  # 4
```

```python
# for percorre os elementos
for preco in precos:
    print(preco)
```

```python
# Menor valor até o momento
temperaturas = [22, 19, 25, 17, 21]
menor = temperaturas[0]

for t in temperaturas:
    if t < menor:
        menor = t

print(menor)  # 17
```

## Erros comuns

| Erro | Exemplo | Causa |
|---|---|---|
| `IndexError` | `precos[4]` em uma lista de 4 elementos | As posições vão de 0 a 3 |
| `SyntaxError` | `[1, 2 3]` | Falta vírgula entre os elementos |
| Erro de lógica | `maior = 0` | Com `[-5, -2, -9]`, imprime 0, que nem está na lista |
| `TypeError` | `max = 10` e, depois, `max([1, 2])` | O nome `max` passou a guardar um número, e não a função |
