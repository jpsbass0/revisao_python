# Módulo 3.2 — Posições, `range()` e `break`: a quinta vogal

**Objetivos de aprendizagem**

- Acessar caracteres pela posição (`palavra[i]`).
- Percorrer posições com `range(len(...))`.
- Interromper um loop com `break` e usar variável de controle.

**Pré-requisitos:** Módulo 3.1.

## Conceitos

- **`palavra[i]`:** acessa o caractere na posição `i`, começando em `0`.
- **Posições válidas:** vão de `0` a `len(palavra) - 1`.
- **`range(n)`:** gera os números de `0` a `n - 1`.
- **`range(len(palavra))`:** gera todas as posições da palavra.
- **`break`:** encerra o loop na hora.
- **Variável de controle:** registra se algo foi encontrado, para usar depois do loop.
- **`not`:** inverte um `bool`.

| Loop | O que `i` guarda | Quando usar |
|---|---|---|
| `for i in palavra` | A letra | Quando só importa a letra |
| `for i in range(len(palavra))` | A posição | Quando importa onde a letra está |

## Exemplos

Rode as células em ordem, de cima para baixo.

```python
# Posições
palavra = "banana"

print(palavra[0])   # b
print(palavra[2])   # n
print(len(palavra)) # 6
```

```python
# Percorrer pela posição
palavra = "ana"

for i in range(len(palavra)):
    print(i, palavra[i])  # 0 a, 1 n, 2 a (um par por linha)
```

```python
# break: para na primeira ocorrência
palavra = "banana"

for i in range(len(palavra)):
    if palavra[i] == "n":
        print(i)  # 2
        break
```

```python
# Variável de controle
palavra = "banana"
encontrou = False

for i in range(len(palavra)):
    if palavra[i] == "z":
        encontrou = True
        break

if not encontrou:
    print("Não encontrei")
```

## Erros comuns

| Erro | Exemplo | Causa |
|---|---|---|
| `IndexError` | `"banana"[6]` | A última posição é 5 |
| `TypeError` | `range("banana")` | `range()` precisa de número; falta `len()` |
| Erro de lógica | Imprimir o contador de vogais | A atividade pede a posição (`i`), não a contagem |
| Erro de lógica | Testar `"Não encontrei"` dentro do loop | A mensagem aparece a cada volta |
