# Módulo 3.4 — Validação de entrada com `while True`

**Objetivos de aprendizagem**

- Repetir uma pergunta até a entrada ser válida, com `while True` e `break`.
- Combinar condições com `and` e `or`.
- Validar números e opções em texto antes de usá-los.

**Pré-requisitos:** Módulo 3.3.

## Conceitos

- **Validação com `while True`:** o `input()` fica dentro do loop. Se a entrada for válida, `break` encerra o loop; se não for, o loop pergunta de novo.
- **Condição do `if`:** descreve a entrada válida, porque é ela que leva ao `break`.
- **`and`:** é verdadeiro só se as duas condições forem verdadeiras.
- **`or`:** é verdadeiro se pelo menos uma condição for verdadeira.
- **`.isdecimal()`:** retorna `True` se o texto tiver pelo menos um caractere e só algarismos. `"7"` dá `True`; `"abc"`, `"-1"`, `"7.5"` e `""` dão `False`.
- **Curto-circuito do `and`:** se a primeira condição for `False`, a segunda não é avaliada. Em `texto.isdecimal() and int(texto) <= 10`, o `int()` só roda quando o texto tem só algarismos, então não gera `ValueError`.
- **Depois do loop:** a variável guarda a última entrada lida, que é a válida.

## Exemplos

Rode as células em ordem, de cima para baixo.

```python
# and e or
idade = 20

print(idade >= 18 and idade < 65)  # True
print(idade < 18 or idade >= 65)   # False
```

```python
# isdecimal(): o texto tem só algarismos?
print("7".isdecimal())    # True
print("10".isdecimal())   # True
print("abc".isdecimal())  # False
print("-1".isdecimal())   # False
print("7.5".isdecimal())  # False
print("".isdecimal())     # False
```

```python
# Curto-circuito: o int() não roda, porque a primeira condição é False
texto = "abc"

print(texto.isdecimal() and int(texto) > 0)  # False, sem erro
```

```python
# Validação de opção em texto com while True
while True:
    resposta = input("Continuar? (s/n): ")
    if resposta == "s" or resposta == "n":
        break
    print("Resposta inválida.")

print(f"Você respondeu: {resposta}")
```

## Erros comuns

| Erro | Exemplo | Causa |
|---|---|---|
| `ValueError` | `int(input("Número: "))` com a entrada `abc` | Conversão antes de validar o texto |
| `ValueError` | `int(texto) <= 10 and texto.isdecimal()` com a entrada `abc` | Ordem invertida: o `int()` roda antes da checagem |
| `SyntaxError` | `if x < 0 or > 10:` | Falta repetir a variável depois do `or` |
| Loop infinito | `input()` antes do `while True` | O valor nunca muda; se for inválido, o loop não para |
| Loop infinito | `if resposta == "s" and resposta == "n":` | Sempre falso, porque a resposta nunca é igual às duas ao mesmo tempo; o `break` nunca roda |
| Erro de lógica | `break` fora do `if` | O loop para na primeira volta e aceita qualquer entrada |
