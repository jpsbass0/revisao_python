# Módulo 6.4 — Descompressão: repetir a letra pelo número

**Objetivos de aprendizagem**

- Separar letras de algarismos com `.isdecimal()`.
- Juntar algarismos seguidos em um número de vários dígitos.
- Repetir um texto com o operador `*`.

**Pré-requisitos:** Módulo 6.3.

## Conceitos

- **Descompressão:** operação inversa da compressão: `ab3c2` volta a ser `abbbcc`. Letra sem número aparece uma vez.
- **`.isdecimal()`:** visto no Módulo 3.4. Em um único caractere, diz se ele é um algarismo.
- **Número com vários algarismos:** `a12` significa 12 letras `a`. Acumule os algarismos em um texto (`"1"`, depois `"12"`) e converta com `int()` só quando o bloco terminar.
- **`*` com texto:** `"a" * 3` dá `"aaa"`. O número de repetições precisa ser `int`.
- **Fechar o bloco:** o bloco termina quando aparece uma nova letra ou quando o texto acaba, como nos Módulos 6.2 e 6.3.

## Exemplos

Rode as células em ordem, de cima para baixo.

```python
# * repete texto
print("a" * 3)         # aaa
print("xy" * 2)        # xyxy
print("b" * int("4"))  # bbbb
```

```python
# isdecimal() em cada caractere
for caractere in "w4":
    print(caractere, caractere.isdecimal())
```

```python
# Juntar os algarismos antes de converter
algarismos = ""

for caractere in "125":
    algarismos += caractere

print(int(algarismos) + 1)  # 126
```

## Erros comuns

| Erro | Exemplo | Causa |
|---|---|---|
| `TypeError` | `"a" * "3"` | O número de repetições precisa ser `int`; converta com `int()` |
| `ValueError` | `int("")` | Letra sem número não tem algarismos para converter; ela aparece uma vez |
| Erro de lógica | Converter cada algarismo assim que ele aparece | `a12` vira 1 e 2 repetições, e não 12 |
| Erro de lógica | Não fechar o bloco depois do `for` | A última letra, como em `m3`, some |
