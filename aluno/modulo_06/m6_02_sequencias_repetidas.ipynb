# Módulo 6.2 — Blocos de caracteres: as sequências repetidas

**Objetivos de aprendizagem**

- Dividir uma string em blocos de caracteres iguais e seguidos.
- Detectar o fim de um bloco quando o caractere muda.
- Tratar o último bloco, que termina junto com o loop.

**Pré-requisitos:** Módulo 6.1.

## Conceitos

- **Bloco:** trecho de caracteres iguais e seguidos. `abcdddxx` tem os blocos `a`, `b`, `c`, `ddd` e `xx`.
- **Sequência de caracteres repetidos:** bloco com 2 ou mais caracteres.
- **Bloco atual:** acumulador de texto com o bloco em formação. Se o caractere for igual ao último do bloco (`bloco[-1]`), ele entra no bloco; se for diferente, o bloco terminou, e um novo começa com esse caractere.
- **Último bloco:** nenhum caractere diferente aparece depois dele, então ele só termina com o fim do loop. Trate-o depois do `for`.
- **Bloco vazio:** `bloco[-1]` gera `IndexError` se `bloco` for `""`. O curto-circuito também vale para o `or`: em `bloco == "" or caractere == bloco[-1]`, se a primeira condição for `True`, a segunda não é avaliada.

## Exemplos

Rode as células em ordem, de cima para baixo.

```python
# Último caractere de um acumulador
bloco = "ddd"

print(bloco[-1])         # d
print("d" == bloco[-1])  # True
```

```python
# Curto-circuito do or: bloco[-1] não roda quando o bloco está vazio
bloco = ""

print(bloco == "" or bloco[-1] == "a")  # True, sem erro
```

```python
# Quantos blocos tem "aabccc"?
texto = "aabccc"
bloco = ""
quantidade = 0

for caractere in texto:
    if bloco == "" or caractere == bloco[-1]:
        bloco += caractere
    else:
        quantidade += 1
        bloco = caractere

# O último bloco termina com o loop
if bloco != "":
    quantidade += 1

print(quantidade)  # 3
```

## Erros comuns

| Erro | Exemplo | Causa |
|---|---|---|
| `IndexError` | `caractere == bloco[-1]` com `bloco = ""` | Texto vazio não tem último caractere; teste `bloco == ""` antes |
| Erro de lógica | Não tratar o bloco depois do `for` | A última sequência, como `mm` em `aymm`, fica de fora |
| Erro de lógica | `if len(bloco) > 2:` | Deixa de fora as sequências de 2 caracteres |
| Erro de lógica | Esquecer `bloco = caractere` ao trocar de bloco | O caractere que abre o novo bloco se perde |
