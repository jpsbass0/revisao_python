# Módulo 6.3 — Compressão: a letra e a quantidade de repetições

**Objetivos de aprendizagem**

- Contar o tamanho de cada bloco de letras repetidas.
- Converter números em texto com `str()`.
- Montar um texto novo a partir dos blocos.

**Pré-requisitos:** Módulo 6.2.

## Conceitos

- **Compressão por repetição:** cada bloco de letras repetidas vira a letra seguida da quantidade. Um bloco de uma letra só fica como está: `abbbcc` vira `ab3c2`.
- **Letra atual e contador:** em vez de guardar o bloco inteiro (Módulo 6.2), guarde a letra do bloco e quantas vezes ela apareceu.
- **`str()`:** converte um número em texto. `"b" + 3` gera `TypeError`; `"b" + str(3)` dá `"b3"`.
- **Fechar o bloco:** quando a letra muda, acrescente o bloco anterior ao resultado e recomece a contagem em 1. O último bloco é fechado depois do loop.
- **Quantidade com dois algarismos:** 12 letras iguais viram `a12`; o `str()` cuida disso.
- **Código repetido:** a regra de fechar o bloco aparece duas vezes, dentro e depois do loop. No Módulo 7, funções evitam esse tipo de repetição.

## Exemplos

Rode as células em ordem, de cima para baixo.

```python
# str() converte número em texto
quantidade = 3

print("b" + str(quantidade))  # b3
print(str(12) + "x")          # 12x
```

```python
# Letra atual e contador: o tamanho de cada bloco de "aaab"
texto = "aaab"
atual = texto[0]
contagem = 0

for letra in texto:
    if letra == atual:
        contagem += 1
    else:
        print(atual, contagem)
        atual = letra
        contagem = 1

# O último bloco termina com o loop
print(atual, contagem)  # b 1
```

## Erros comuns

| Erro | Exemplo | Causa |
|---|---|---|
| `TypeError` | `resultado += atual + contagem` | Não se soma texto com número; use `str(contagem)` |
| Erro de lógica | Acrescentar sempre a quantidade | Gera `a1b3` em vez de `ab3` |
| Erro de lógica | Não fechar o bloco depois do `for` | O último bloco, como `m3`, some |
| Erro de lógica | `contagem = 0` ao começar um bloco novo | A letra que abre o bloco não é contada |
