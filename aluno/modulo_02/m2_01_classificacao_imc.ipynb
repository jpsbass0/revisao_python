## Conceitos

- **`bool`:** tipo com dois valores, `True` e `False`.
- **Comparações:** `==` (igual), `!=` (diferente), `<`, `<=`, `>` e `>=`. O resultado é sempre um `bool`.
- **`if`:** executa um bloco se a condição for verdadeira.
- **`else`:** cobre todos os casos restantes e não leva condição.
- **`elif`:** só é testado se as condições anteriores forem falsas. Em uma cadeia `if`/`elif`/`else`, roda apenas o primeiro bloco verdadeiro. Vários `if` separados são todos testados.
- **Sintaxe:** os `:` e o recuo são obrigatórios. O recuo define o que está dentro do bloco.
- **`=` x `==`:** `=` atribui um valor; `==` compara dois valores.

## Exemplos

Rode as células em ordem, de cima para baixo.

```python
# Comparações retornam True ou False
preco = 38.50

print(preco > 30)     # True
print(preco == 40)    # False
print(preco != 40)    # True
print(preco <= 38.5)  # True
```

```python
# if sem else: se a condição for falsa, o bloco é ignorado
saldo = 100

if saldo < 0:
    print("Conta negativa")

print("Fim")  # Fim
```

```python
# if/else (dado fictício)
retorno = -0.02

if retorno >= 0:
    print("Ganho")
else:
    print("Perda")  # Perda
```

```python
# if/elif/else: rode com as notas 8, 5 e 3
nota = float(input("Nota: "))

if nota >= 7:
    print("Aprovado")
elif nota >= 4:
    print("Prova final")
else:
    print("Reprovado")
```

```python
# O recuo define o bloco
nota = 5

if nota >= 7:
    print("Aprovado")   # Dentro do if: não roda
    print("Parabéns")   # Dentro do if: não roda

print("Fim")            # Fora do if: sempre roda
```

```python
# A ordem importa: este código está errado de propósito
nota = 8

if nota >= 4:
    print("Prova final")  # Roda, porque 8 >= 4
elif nota >= 7:
    print("Aprovado")     # Nunca é testado para a nota 8
```

## Erros comuns

| Erro | Exemplo | Causa |
|---|---|---|
| `SyntaxError` | `if nota >= 7` | Falta `:` no fim da linha |
| `SyntaxError` | `if nota = 7:` | Uso de `=` no lugar de `==` |
| `SyntaxError` | `else nota < 4:` | `else` com condição |
| `IndentationError` | `print()` sem recuo logo abaixo do `if` | Bloco sem recuo |
| `TypeError` | `input("Nota: ") >= 7` | Comparação de texto com número; falta converter com `float()` |
