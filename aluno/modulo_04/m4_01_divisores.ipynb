# Módulo 4.1 — `range()` e resto da divisão: contar divisores

**Objetivos de aprendizagem**

- Gerar sequências de inteiros com `range()`.
- Testar divisibilidade com o operador `%`.
- Combinar `for`, `if` e contador para contar ocorrências.

**Pré-requisitos:** Módulo 3.5.

## Conceitos

- **`range(inicio, fim)`:** gera os inteiros de `inicio` até `fim - 1`; o `fim` fica de fora. `range(1, 5)` gera 1, 2, 3 e 4. Com um só argumento, começa em 0: `range(3)` gera 0, 1 e 2.
- **`for` com `range()`:** `for i in range(1, n + 1):` percorre os inteiros de 1 até `n`, incluindo `n`.
- **`%` (resto):** `a % b` é o resto da divisão de `a` por `b`. Se o resto for `0`, `b` divide `a` exatamente.
- **Divisor:** `d` é divisor de `n` quando `n % d == 0`. Os divisores positivos de um inteiro positivo `n` estão entre 1 e `n`.
- **Contador:** o mesmo do Módulo 3.1, agora aumentado quando o teste de divisibilidade dá verdadeiro.

## Exemplos

Rode as células em ordem, de cima para baixo.

```python
# range(): o fim fica de fora
for i in range(1, 5):
    print(i)
```

```python
# Resto da divisão
print(10 % 3)  # 1
print(10 % 5)  # 0
print(7 % 2)   # 1
```

```python
# Divisível ou não?
numero = 15

print(numero % 3 == 0)  # True
print(numero % 4 == 0)  # False
```

```python
# Contar os números pares de 1 a 10
pares = 0

for i in range(1, 11):
    if i % 2 == 0:
        pares += 1

print(pares)  # 5
```

## Erros comuns

| Erro | Exemplo | Causa |
|---|---|---|
| Erro de lógica | `range(1, n)` | Deixa o próprio `n` de fora; use `range(1, n + 1)` |
| `ZeroDivisionError` | `range(0, n + 1)` com o teste `n % i` | O primeiro `i` é 0, e o resto da divisão por zero gera erro |
| `TypeError` | `range(1, input("Número: "))` | `input()` retorna `str`; converta com `int()` |
| `TypeError` | `range(1, 7.5)` | `range()` só aceita inteiros |
| Erro de lógica | `if n / i == 0:` | Testa o resultado da divisão, não o resto; use `%` |
| Erro de lógica | `print()` dentro do `for` | Imprime a contagem parcial a cada volta |
