**Pré-requisitos:** Módulo 7.1.

## Conceitos

- **Vários `return`:** uma função pode ter um `return` em cada ramo de um `if`/`elif`/`else`. O primeiro que rodar encerra a função.
- **Função que retorna texto:** `return "Peso normal"` devolve uma `str`, como qualquer outro valor.
- **Combinar funções:** o resultado de uma função pode ser argumento de outra: `classificar_imc(calcular_imc(70, 1.75))`.
- **Testar a função isoladamente:** chamar a função com os valores das fronteiras (18.5, 25 e 30) mostra se os `<` e `<=` estão certos, sem precisar digitar peso e altura.
- **Cada notebook é independente:** a função do Módulo 7.1 não existe aqui até você copiar o `def` e rodar a célula.

## Exemplos

Rode as células em ordem, de cima para baixo.

```python
# Vários return: o primeiro que rodar encerra a função
def situacao(nota):
    if nota >= 7:
        return "Aprovado"
    elif nota >= 4:
        return "Prova final"
    else:
        return "Reprovado"

print(situacao(8))  # Aprovado
print(situacao(5))  # Prova final
print(situacao(3))  # Reprovado
```

```python
# O resultado de uma função como argumento de outra
def media(a, b):
    return (a + b) / 2

print(situacao(media(6, 9)))  # Aprovado
```

## Erros comuns

| Erro | Exemplo | Causa |
|---|---|---|
| Erro de lógica | `if imc <= 25:` no teste de peso normal | O IMC 25 deveria ser `Sobrepeso`, mas cai em `Peso normal` |
| Erro de lógica | `print("Peso normal")` no lugar do `return` | A função devolve `None`, e a classificação não chega ao `print()` final |
| `NameError` | `calcular_imc` usada sem o `def` neste notebook | Funções de outros notebooks não existem aqui; copie o `def` |
| `TypeError` | `classificar_imc()` | Falta o argumento com o valor do IMC |
