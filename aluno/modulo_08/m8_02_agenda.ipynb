# Módulo 8.2 — Dicionário em modo contínuo: agenda telefônica

**Objetivos de aprendizagem**

- Incluir dados em um dicionário até o usuário decidir parar.
- Atualizar um valor que depende do valor anterior da mesma chave.
- Exibir um dicionário com `for`.

**Pré-requisitos:** Módulos 3.3 e 8.1.

## Conceitos

- **Modo contínuo:** um `while True` lê dados e os inclui no dicionário até o usuário escolher parar. Aqui, a escolha é pressionar Enter sem digitar o nome, o texto vazio do Módulo 3.3.
- **Nome como chave, telefone como valor:** `agenda[nome] = telefone`.
- **Telefone como texto:** guarde o telefone como `str`. Ele pode ter hífen, parênteses ou zero à esquerda e não entra em contas.
- **Chave repetida:** incluir de novo um nome que já existe substitui o telefone antigo.
- **Atualizar a partir do valor atual:** `votos[nome] += 1` só funciona se a chave já existir; teste com `in` antes.
- **Exibir:** `for nome in agenda:` percorre as chaves na ordem de inclusão.

## Exemplos

Rode as células em ordem, de cima para baixo.

```python
# Chave repetida substitui o valor
precos = {}
precos["café"] = 5.0
precos["pão"] = 1.5
precos["café"] = 6.0

print(precos)  # {'café': 6.0, 'pão': 1.5}
```

```python
# Exibir com for
for produto in precos:
    print(f"{produto}: R$ {precos[produto]:.2f}")
```

```python
# Modo contínuo: contar votos até o Enter vazio
votos = {}

while True:
    candidato = input("Voto (Enter vazio para encerrar): ")
    if candidato == "":
        break
    if candidato in votos:
        votos[candidato] += 1
    else:
        votos[candidato] = 1

print(votos)
```

## Erros comuns

| Erro | Exemplo | Causa |
|---|---|---|
| `KeyError` | `votos[candidato] += 1` sem testar `in` | No primeiro voto, a chave ainda não existe |
| Erro de lógica | `agenda = {}` dentro do `while` | A agenda volta a ficar vazia a cada contato |
| Erro de lógica | Pedir o telefone antes de testar o nome vazio | O usuário precisa digitar um telefone só para sair |
| `ValueError` | `int(input("Telefone: "))` com a entrada `0000-0001` | O hífen não é algarismo; guarde o telefone como texto |
