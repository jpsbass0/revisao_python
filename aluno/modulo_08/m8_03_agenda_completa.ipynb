# Módulo 8.3 — Listas como valores: agenda completa

**Objetivos de aprendizagem**

- Guardar vários dados por chave, em uma lista.
- Acessar um item da lista que está dentro do dicionário.
- Organizar o programa em funções que recebem e retornam o dicionário.

**Pré-requisitos:** Módulo 8.2.

## Conceitos

- **Lista como valor:** `agenda["Ana"] = ["0000-0001", "ana@exemplo.com", "Rio de Janeiro"]` guarda vários dados para a mesma chave.
- **Acesso em dois passos:** `agenda["Ana"]` é a lista; `agenda["Ana"][1]` é o segundo item dela, o e-mail.
- **Ordem fixa dos dados:** todos os contatos guardam os dados na mesma ordem (telefone, e-mail e cidade), para que cada posição sempre signifique a mesma coisa.
- **Função que retorna o dicionário:** a função que lê os contatos monta a agenda e a devolve com `return`.
- **Função que recebe o dicionário:** a função que exibe a agenda a recebe como parâmetro.

## Exemplos

Rode as células em ordem, de cima para baixo.

```python
# Lista como valor e acesso em dois passos
alunos = {"Ana": [8.5, "Administração"], "Bia": [9.0, "Economia"]}

print(alunos["Ana"])     # [8.5, 'Administração']
print(alunos["Ana"][1])  # Administração
```

```python
# Incluir um par com lista e exibir todos
alunos["Caio"] = [7.0, "Direito"]

for nome in alunos:
    dados = alunos[nome]
    print(f"{nome}: nota {dados[0]} | curso {dados[1]}")
```

```python
# Função que recebe o dicionário como parâmetro
def exibir(dicionario):
    for chave in dicionario:
        print(chave, dicionario[chave])

exibir({"x": [1, 2], "y": [3]})
```

## Erros comuns

| Erro | Exemplo | Causa |
|---|---|---|
| `IndexError` | `agenda["Ana"][3]` com 3 dados por contato | As posições da lista vão de 0 a 2 |
| Erro de lógica | Ordem dos dados diferente entre contatos | A posição 1 deixa de ser sempre o e-mail |
| Erro de lógica | `dados = []` antes do `while`, com `.append()` de cada dado | A mesma lista recebe os dados de todos os contatos |
| `TypeError` | `exibir_agenda()` | A função espera a agenda como argumento |
