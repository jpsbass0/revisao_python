**Objetivos de aprendizagem**

- Criar dicionários e acessar valores pelas chaves.
- Incluir pares chave-valor dentro de um loop.
- Usar listas como valores de um dicionário.

**Pré-requisitos:** Módulo 7.5.

## Conceitos

- **Dicionário:** guarda pares `chave: valor` entre chaves: `{"Ana": 20, "Bia": 22}`. O valor é acessado pela chave, e não pela posição: `idades["Ana"]` dá `20`.
- **Dicionário vazio:** `{}`.
- **Incluir ou alterar:** `dicionario[chave] = valor` cria o par se a chave não existir e substitui o valor se ela já existir. As chaves não se repetem.
- **`in`:** `"Ana" in idades` testa se a chave existe.
- **Ordem:** o dicionário mantém a ordem em que as chaves foram incluídas.
- **`for` em dicionário:** `for chave in dicionario:` percorre as chaves; o valor de cada uma é `dicionario[chave]`.
- **Valores podem ser listas:** `{4: [1, 2, 4]}` associa à chave `4` a lista dos seus divisores. Uma função pode montar a lista (com `.append()`, do Módulo 5.3) e retorná-la.

## Exemplos

Rode as células em ordem, de cima para baixo.

```python
# Criar, acessar, incluir e alterar
idades = {"Ana": 20, "Bia": 22}
print(idades["Ana"])  # 20

idades["Caio"] = 19  # inclui
idades["Ana"] = 21   # altera
print(idades)          # {'Ana': 21, 'Bia': 22, 'Caio': 19}
print(len(idades))     # 3
print("Bia" in idades)  # True
```

```python
# Montar um dicionário em um loop: cada número e o seu quadrado
quadrados = {}

for n in range(1, 5):
    quadrados[n] = n ** 2

print(quadrados)  # {1: 1, 2: 4, 3: 9, 4: 16}
```

```python
# Valores que são listas
notas = {"Ana": [8, 9], "Bia": [6]}
notas["Bia"].append(7)

print(notas["Bia"])  # [6, 7]
for nome in notas:
    print(nome, notas[nome])
```

```python
# Função que monta e retorna uma lista
def multiplos(n, quantidade):
    lista = []
    for i in range(1, quantidade + 1):
        lista.append(n * i)
    return lista

print(multiplos(3, 4))  # [3, 6, 9, 12]
```

## Erros comuns

| Erro | Exemplo | Causa |
|---|---|---|
| `KeyError` | `idades["Duda"]` | A chave não existe no dicionário |
| `KeyError` | `idades[0]` | Dicionário não tem posições; o acesso é pela chave |
| `TypeError` | `{[1, 2]: "x"}` | Lista não pode ser chave; números e strings podem |
| Erro de lógica | `resultado = {}` dentro do `for` | O dicionário volta a ficar vazio a cada volta e só guarda o último número |
